```python
#ejercicio 1
text = """Beautiful is better than ugly.
Explicit is better than implicit.
Simple is better than complex.
Complex is better than complicated.
Flat is better than nested.
Sparse is better than dense.
Readability counts.
Special cases aren't special enough to break the rules.
Although practicality beats purity.
Errors should never pass silently.
Unless explicitly silenced.
In the face of ambiguity, refuse the temptation to guess.
There should be one-- and preferably only one --obvious way to do it.
Although that way may not be obvious at first unless you're Dutch.
Now is better than never.
Although never is often better than *right* now.
If the implementation is hard to explain, it's a bad idea.
If the implementation is easy to explain, it may be a good idea.
Namespaces are one honking great idea -- let's do more of those!"""

lineas = text.splitlines()
cantidad_lineas = len(text.splitlines())
cantidad_palabras = len(text.split())
promedio = cantidad_palabras / cantidad_lineas
print(f"Total de lineas: {cantidad_lineas}")
print(f"Total de palabras: {cantidad_palabras}")
print(f"Promedio de palabras por linea: {promedio:.2f}")
print(f"\nLineas por encima del promedio ({promedio:.2f} palabras):")

for linea in lineas:
  if len(linea.split()) > promedio:
    print(f'- "{linea}" ({len(linea.split())} palabras)')
```

```python
#ejercicio2
playlist = [
{"title": "Bohemian Rhapsody", "duration": "5:55"},
{"title": "Hotel California", "duration": "6:30"},
{"title": "Stairway to Heaven", "duration": "8:02"},
{"title": "Imagine", "duration": "3:07"},
{"title": "Smells Like Teen Spirit", "duration": "5:01"},
{"title": "Billie Jean", "duration": "4:54"},
{"title": "Hey Jude", "duration": "7:11"},
{"title": "Like a Rolling Stone", "duration": "6:13"},
]

total_segundos = 0
mas_larga = None
mas_corta = None

for cancion in playlist:
  minutos, segundos = cancion["duration"].split(":")
  duracion_total = int(minutos) * 60 + int(segundos)

  total_segundos+= duracion_total

  if mas_larga is None or duracion_total > mas_larga[1]:
    mas_larga = (cancion["title"],duracion_total)

  if mas_corta is None or duracion_total < mas_corta[1]:
    mas_corta = (cancion["title"], duracion_total)

minutos_totales = total_segundos // 60
segundos_totales = total_segundos % 60

print(f"Duracion total: {minutos_totales}m {segundos_totales}s")
print(f'Cancion mas larga: "{mas_larga[0]}" ({mas_larga[1] // 60},{mas_larga[1] % 60:02d})')
print(f'Cancion mas corta: "{mas_corta[0]}" ({mas_corta[1] // 60},{mas_corta[1] % 60:02d})')
```

```python
#ejercicio3
review = """La película sigue a un grupo de astronautas que viajan a Marte en una misión de rescate.
El capitán Torres lidera al equipo a través de tormentas solares y fallos en el sistema de navegación.
Al llegar a Marte descubren que la base está abandonada y los suministros destruidos.
Torres decide sacrificar la nave nodriza para salvar al equipo y logran volver a la Tierra en una cápsula de emergencia.
El final revela que Torres sobrevivió gracias a un pasaje secreto."""

entrada = input("Ingrese las palabras spoilers separadas por coma: ")
palabras_usuario = [p.strip().lower() for p in entrada.split(",")]
lineas = review.splitlines()

for i, linea in enumerate(lineas):
  palabras_linea = linea.split()

  for j, palabra in enumerate(palabras_linea):
    if palabra.lower() in palabras_usuario:
      palabras_linea[j] = "*" * len(palabra)

  lineas[i] = " ".join(palabras_linea)

review_final = "\n".join(lineas)

print(review_final)
```

```python
#ejercicio4
entrada = input("Ingrese email: ")
esValido = False
if entrada.count("@") == 1:
  primer_parametro, segundo_parametro = entrada.split("@")
  if len(primer_parametro) > 0 and "." in segundo_parametro and not segundo_parametro.startswith("."):
    if not entrada.startswith("@") and not entrada.startswith(".") and not entrada.endswith("@") and not entrada.endswith("."):
        ultimo_punto = segundo_parametro.rfind(".")
        dominio = segundo_parametro[ultimo_punto+1:]
        if len(dominio) > 1:
          esValido = True
if esValido:
  print("El email es valido")
else:
  print("El email no es valido")
```

```python
#ejercicio5
precios = {
    "local": {"hasta 1": 500, "hasta 5": 1000, "más de 5": 2000},
    "regional": {"hasta 1": 1000, "hasta 5": 2500, "más de 5": 5000},
    "nacional": {"hasta 1": 2000, "hasta 5": 4500, "más de 5": 8000}
}
peso = input("ingrese peso del paquete: ")
zona = input("Ingrese zona de destino (local/regional/nacional): ").lower()
if float(peso) <= 1:
    peso = "hasta 1"
elif float(peso) <= 5:
    peso = "hasta 5"
else:   peso = "más de 5"

if zona in precios:
    if peso in precios[zona]:
        costo = precios[zona][peso]
        print(f"El costo de envio es: {costo}")
    else:
        print("Peso no valido")
else:   print("Zona no valida. Las zonas validas son: local, regional, nacional")
```

```python
#ejercicio6 
posts = [
"Arrancando el lunes con energía #Motivación #NuevaSemana", "Terminé mi primer proyecto en Python #Python #Programación #OrgullosoDeMi", "No puedo creer el final de la serie #SinSpoilers #SerieAdicta","Nuevo video en el canal sobre #InteligenciaArtificial y #Python", "Entrenamiento de hoy completado #Fitness #Motivación #NoPainNoGain", "Leyendo sobre #InteligenciaArtificial y el futuro del trabajo #Tecnología", "Arranqué a estudiar #Programación por mi cuenta #Python #Autodidacta", "Finde de lluvia, maratón de series #SerieAdicta #Relax", "Workshop de #InteligenciaArtificial en la universidad #Tecnología #Programación"
]

hashtags = {}

for post in posts:
  palabras = post.split()
  for palabra in palabras:
    if palabra.startswith("#"):
      hashtag = palabra[1:]
      if hashtag in hashtags:
        hashtags[hashtag] += 1
      else:
        hashtags[hashtag] = 1

ordenado = dict(sorted(hashtags.items(), key=lambda x: x[1], reverse=True))
hashtags_unicos = 0
print("Hashtags trending (más de una aparición)")
for hashtag, cantidad in ordenado.items():
  if cantidad > 1:
    print(f"#{hashtag}: {cantidad} apariciones")
  hashtags_unicos+=1
print(f"\nHashtags únicos (solo una aparición): {hashtags_unicos}")
```

```python
#ejercicio7
import random

entrada = [p.strip() for p in input("Ingrese nombres de participantes separados por coma: ").lower().split(",")]
participantes = set(entrada) #Elimina duplicados.
if len(participantes) > 2: #Verifica que haya al menos 3 participantes.
    participantes = random.sample(list(participantes), len(participantes))
    for i in range(0, len(participantes),2):
        if i+1 < len(participantes):
            print(f"{participantes[i]} -> {participantes[i+1]}")
        else:
            print(f"{participantes[i]} sin pareja.")
else: print("Se necesitan al menos 3 participantes para realizar el sorteo.")
```
